```python
#W9A1
def merge_sorted_lists(list1, list2):
    i, j = 0, 0
    res = []
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            res.append(list1[i])
            i += 1
        else:
            res.append(list2[j])
            j += 1
    res.extend(list1[i:])
    res.extend(list2[j:])
    return res
```

```python
#W9A2
import heapq

def kth_largest_matrix(matrix, k):
    n, m = len(matrix), len(matrix[0])
    min_heap = []
    for r in range(n):
        for c in range(m):
            if len(min_heap) < k:
                heapq.heappush(min_heap, matrix[r][c])
            else:
                heapq.heappushpop(min_heap, matrix[r][c])
    return min_heap[0]
```

```python
#W9A3
def swap_smallest(n):
    s = list(str(n))
    for i in range(len(s)):
        min_idx = i
        for j in range(len(s) - 1, i, -1):
            if s[j] < s[min_idx]:
                if i != 0 or s[j] != '0':
                    min_idx = j
        if min_idx != i:
            s[i], s[min_idx] = s[min_idx], s[i]
            return int("".join(s)) 
    if len(s) >= 2:
        s[-1], s[-2] = s[-2], s[-1]
    return int("".join(s))
```

```python
#W9A4
def can_transform(a, b):
    dp = [[False] * (len(b) + 1) for _ in range(len(a) + 1)]
    dp[0][0] = True 
    for i in range(1, len(a) + 1):
        dp[i][0] = dp[i-1][0] and a[i-1].islower()
        for j in range(1, len(b) + 1):
            if a[i-1].islower():
                dp[i][j] = dp[i-1][j] or (dp[i-1][j-1] and a[i-1].upper() == b[j-1])
            else:
                dp[i][j] = dp[i-1][j-1] and a[i-1] == b[j-1]     
    return "YES" if dp[len(a)][len(b)] else "NO"
```

```python
#W9A5
def is_sublist(list1, list2):
    n, m = len(list1), len(list2)
    if n == 0: 
        return "YES"
    for i in range(m - n + 1):
        if list2[i:i+n] == list1:
            return "YES"
    return "NO"
```

```python
#W9A6
def next_permutation(arr):
    n = len(arr)
    i = n - 2
    while i >= 0 and arr[i] >= arr[i + 1]:
        i -= 1
    if i >= 0:
        j = n - 1
        while arr[j] <= arr[i]:
            j -= 1
        arr[i], arr[j] = arr[j], arr[i]   
    arr[i + 1:] = reversed(arr[i + 1:])
    return arr
```

```python
#W9A7
def canPlaceFlowers(flowerbed, k):
    count = 0
    n = len(flowerbed)
    for i in range(n):
        if flowerbed[i] == 0:
            empty_left = (i == 0) or (flowerbed[i - 1] == 0)
            empty_right = (i == n - 1) or (flowerbed[i + 1] == 0)           
            if empty_left and empty_right:
                flowerbed[i] = 1
                count += 1
                
    return count >= k
```

```python
#W9A8
def is_path_crossing(moves):
    x, y = 0, 0
    visited = {(0, 0)}
    for move in moves:
        if move == 'R': x += 1
        elif move == 'L': x -= 1
        elif move == 'U': y += 1
        elif move == 'D': y -= 1
        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False
```

```python
#W9A9
def Median(nums):
    a, b, c, d, e = nums[0], nums[1], nums[2], nums[3], nums[4]
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c:
        b, d = d, b
        c = a
    if e < c: e, c = c, e
    if e < b: e, b = b, e
    if b < c: return c
    return b
```

```python
#W9A10
import bisect
def findRadius(houses, heaters):
    heaters.sort()
    max_radius = 0
    for house in houses:
        idx = bisect.bisect_left(heaters, house)
        left_dist = house - heaters[idx - 1] if idx > 0 else float('inf')
        right_dist = heaters[idx] - house if idx < len(heaters) else float('inf')
        max_radius = max(max_radius, min(left_dist, right_dist))
    return max_radius
```
